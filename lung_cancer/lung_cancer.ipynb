```python
# =========================================================
# 1️⃣ IMPORTS
# =========================================================

import os
import numpy as np
import pandas as pd
import pydicom
import cv2
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, roc_auc_score
from tqdm import tqdm

# =========================================================
# 2️⃣ PATHS
# =========================================================

base_path = r"C:\Users\nandh\OneDrive\Desktop\lung cancer\manifest-1772384640522\NSCLC Radiogenomics"
clinical_path = r"C:\Users\nandh\OneDrive\Desktop\lung cancer\NSCLCR01Radiogenomic_DATA_LABELS_2018-05-22_1500-shifted.csv"
save_model_path = r"C:\Users\nandh\OneDrive\Desktop\lung cancer\lung_survival_model.pth"

# =========================================================
# 3️⃣ LOAD CLINICAL DATA
# =========================================================

clinical = pd.read_csv(clinical_path)

clinical = clinical[['Case ID', 'Survival Status']].copy()

clinical['label'] = clinical['Survival Status'].map({
    'Alive': 0,
    'Dead': 1
})

clinical = clinical.dropna()

label_dict = dict(zip(clinical['Case ID'], clinical['label']))

print("Label Distribution:")
print(clinical['label'].value_counts())

# =========================================================
# 4️⃣ LOAD CT VOLUME FUNCTION
# =========================================================

def load_ct_volume(series_path, target_slices=20, img_size=128):

    slices = []
    files = sorted([f for f in os.listdir(series_path) if f.endswith(".dcm")])

    for file in files:
        try:
            ds = pydicom.dcmread(os.path.join(series_path, file))
            image = ds.pixel_array.astype(np.float32)

            # Normalize
            image = (image - np.min(image)) / (np.max(image) - np.min(image) + 1e-5)

            # Resize
            image = cv2.resize(image, (img_size, img_size))

            slices.append(image)

        except:
            continue

    if len(slices) < target_slices:
        return None

    slices = slices[:target_slices]
    volume = np.stack(slices)

    return volume

# =========================================================
# 5️⃣ BUILD DATASET
# =========================================================

X = []
y = []

patients = os.listdir(base_path)

for patient in tqdm(patients):

    if patient not in label_dict:
        continue

    patient_path = os.path.join(base_path, patient)

    for study in os.listdir(patient_path):

        if "CT" in study.upper():

            study_path = os.path.join(patient_path, study)

            for series in os.listdir(study_path):

                series_path = os.path.join(study_path, series)

                if os.path.isdir(series_path):

                    volume = load_ct_volume(series_path)

                    if volume is not None:
                        X.append(volume)
                        y.append(label_dict[patient])
                        break
            break

X = np.array(X)
y = np.array(y)

print("\nFinal Dataset Shape:", X.shape)
print("Dead count:", sum(y))

# =========================================================
# 6️⃣ CONVERT TO TENSORS + SPLIT
# =========================================================

X = torch.tensor(X, dtype=torch.float32).unsqueeze(1)
y = torch.tensor(y, dtype=torch.float32)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print("Train size:", len(X_train))
print("Test size:", len(X_test))

# =========================================================
# 7️⃣ DEVICE
# =========================================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)

# =========================================================
# 8️⃣ 3D CNN MODEL
# =========================================================

class Survival3DCNN(nn.Module):
    def __init__(self):
        super(Survival3DCNN, self).__init__()

        self.conv = nn.Sequential(
            nn.Conv3d(1, 16, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(16, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool3d(2)
        )

        self.fc = nn.Sequential(
            nn.Flatten(),
            nn.Linear(32 * 5 * 32 * 32, 128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        x = self.conv(x)
        x = self.fc(x)
        return x

model = Survival3DCNN().to(device)

criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=0.0005)

# =========================================================
# 9️⃣ TRAINING
# =========================================================

epochs = 20
best_auc = 0

for epoch in range(epochs):

    model.train()
    optimizer.zero_grad()

    outputs = model(X_train.to(device)).squeeze()
    loss = criterion(outputs, y_train.to(device))

    loss.backward()
    optimizer.step()

    model.eval()
    with torch.no_grad():
        test_outputs = model(X_test.to(device)).squeeze()
        probs = torch.sigmoid(test_outputs).cpu().numpy()
        preds = (probs > 0.5).astype(int)

        acc = accuracy_score(y_test.numpy(), preds)
        auc = roc_auc_score(y_test.numpy(), probs)

    print(f"Epoch {epoch+1}/{epochs} | Loss: {loss.item():.4f} | Acc: {acc:.4f} | AUC: {auc:.4f}")

    if auc > best_auc:
        best_auc = auc
        torch.save(model.state_dict(), save_model_path)
        print("✅ Best model saved!")

print("\nFinal Best AUC:", best_auc)
print("Model saved at:", save_model_path)
```

```python
clinical = pd.read_csv(clinical_path)

clinical = clinical[[
    'Case ID',
    'Survival Status',
    'Age at Histological Diagnosis',
    'Pack Years'
]].copy()

clinical['label'] = clinical['Survival Status'].map({
    'Alive': 0,
    'Dead': 1
})

clinical['Pack Years'] = pd.to_numeric(clinical['Pack Years'], errors='coerce')
clinical = clinical.dropna()

scaler = StandardScaler()
clinical[['Age at Histological Diagnosis', 'Pack Years']] = scaler.fit_transform(
    clinical[['Age at Histological Diagnosis', 'Pack Years']]
)

label_dict = dict(zip(clinical['Case ID'], clinical['label']))
clinical_dict = dict(zip(
    clinical['Case ID'],
    clinical[['Age at Histological Diagnosis', 'Pack Years']].values
))
```

```python
def load_ct_volume(series_path, target_slices=20, img_size=128):

    slices = []
    files = sorted([f for f in os.listdir(series_path) if f.endswith(".dcm")])

    for file in files:
        try:
            ds = pydicom.dcmread(os.path.join(series_path, file))
            image = ds.pixel_array.astype(np.float32)
            image = (image - np.min(image)) / (np.max(image) - np.min(image) + 1e-5)
            image = cv2.resize(image, (img_size, img_size))
            slices.append(image)
        except:
            continue

    if len(slices) < target_slices:
        return None

    return np.stack(slices[:target_slices])
```

```python
X_img = []
X_clin = []
y = []

patients = os.listdir(base_path)

for patient in tqdm(patients):

    if patient not in label_dict:
        continue

    patient_path = os.path.join(base_path, patient)

    for study in os.listdir(patient_path):
        if "CT" in study.upper():

            study_path = os.path.join(patient_path, study)

            for series in os.listdir(study_path):
                series_path = os.path.join(study_path, series)

                if os.path.isdir(series_path):
                    volume = load_ct_volume(series_path)

                    if volume is not None:
                        X_img.append(volume)
                        X_clin.append(clinical_dict[patient])
                        y.append(label_dict[patient])
                        break
            break

X_img = np.array(X_img)
X_clin = np.array(X_clin)
y = np.array(y)

print("Image shape:", X_img.shape)
print("Clinical shape:", X_clin.shape)
print("Dead count:", sum(y))
```

```python
import os
import numpy as np
import pydicom
import cv2
import pandas as pd

X_img = []
X_clin = []
y = []

target_slices = 20
img_size = 128

def load_ct_volume(series_path):

    slices = []

    for file in os.listdir(series_path):
        if file.endswith(".dcm"):
            path = os.path.join(series_path, file)
            try:
                ds = pydicom.dcmread(path)
                slices.append(ds.pixel_array)
            except:
                continue

    if len(slices) < target_slices:
        return None

    # Simply sort by filename (safe fallback)
    slices = sorted(slices, key=lambda x: x.shape[0])

    slices = slices[:target_slices]

    images = []
    for img in slices:
        img = cv2.resize(img, (img_size, img_size))
        img = (img - np.min(img)) / (np.max(img) - np.min(img) + 1e-5)
        images.append(img)

    return np.stack(images)


# Load clinical
clinical = pd.read_csv(clinical_path)

label_dict = {}
for _, row in clinical.iterrows():
    if row['Survival Status'] == 'Dead':
        label_dict[row['Case ID']] = 1
    elif row['Survival Status'] == 'Alive':
        label_dict[row['Case ID']] = 0

print("Total clinical labels:", len(label_dict))


# Build dataset
for patient in os.listdir(base_path):

    if patient not in label_dict:
        continue

    patient_path = os.path.join(base_path, patient)

    if not os.path.isdir(patient_path):
        continue

    found = False

    for study in os.listdir(patient_path):

        study_path = os.path.join(patient_path, study)

        if not os.path.isdir(study_path):
            continue

        for series in os.listdir(study_path):

            series_path = os.path.join(study_path, series)

            if not os.path.isdir(series_path):
                continue

            volume = load_ct_volume(series_path)

            if volume is not None:

                X_img.append(volume)

                age = clinical.loc[
                    clinical['Case ID'] == patient,
                    'Age at Histological Diagnosis'
                ].values[0]

                gender = clinical.loc[
                    clinical['Case ID'] == patient,
                    'Gender'
                ].values[0]

                gender = 1 if gender == 'Male' else 0

                X_clin.append([age, gender])
                y.append(label_dict[patient])

                found = True
                break

        if found:
            break


X_img = np.array(X_img)
X_clin = np.array(X_clin)
y = np.array(y)

print("Final X_img shape:", X_img.shape)
print("Final X_clin shape:", X_clin.shape)
print("Final y shape:", y.shape)
print("Dead count:", sum(y))
```

```python
# =========================================================
# 🔥 FINAL STABLE MULTIMODAL RESEARCH MODEL
# =========================================================

import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
from tqdm import tqdm
from sklearn.metrics import accuracy_score, roc_auc_score
from sklearn.model_selection import train_test_split
from torch.utils.data import Dataset, DataLoader

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# =========================================================
# 1️⃣ ENSURE TENSOR FORMAT
# =========================================================

if not isinstance(X_img, torch.Tensor):
    X_img = torch.tensor(X_img, dtype=torch.float32)

if not isinstance(X_clin, torch.Tensor):
    X_clin = torch.tensor(X_clin, dtype=torch.float32)

if not isinstance(y, torch.Tensor):
    y = torch.tensor(y, dtype=torch.float32)

# Add channel dimension
if X_img.dim() == 4:
    X_img = X_img.unsqueeze(1)

print("Image shape:", X_img.shape)
print("Clinical shape:", X_clin.shape)
print("Label shape:", y.shape)

# =========================================================
# 2️⃣ SAFE RADIOMICS
# =========================================================

def extract_radiomics(volume):
    # volume shape: (20,128,128)
    mean = torch.mean(volume)
    std = torch.std(volume)
    minv = torch.min(volume)
    maxv = torch.max(volume)
    p25 = torch.quantile(volume, 0.25)
    p75 = torch.quantile(volume, 0.75)
    slice_var = torch.var(volume.view(volume.shape[0], -1), dim=1).mean()
    return torch.stack([mean, std, minv, maxv, p25, p75, slice_var])

radiomics = torch.stack([extract_radiomics(v[0]) for v in X_img])

print("Radiomics shape:", radiomics.shape)

# =========================================================
# 3️⃣ TRAIN TEST SPLIT (ALL MODALITIES TOGETHER)
# =========================================================

X_train, X_test, y_train, y_test, rad_train, rad_test, clin_train, clin_test = train_test_split(
    X_img,
    y,
    radiomics,
    X_clin,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# =========================================================
# 4️⃣ DATASET CLASS
# =========================================================

class MultiModalDataset(Dataset):
    def __init__(self, images, clinical, radiomics, labels):
        self.images = images
        self.clinical = clinical
        self.radiomics = radiomics
        self.labels = labels

    def __len__(self):
        return len(self.labels)

    def __getitem__(self, idx):
        return (
            self.images[idx],
            self.clinical[idx],
            self.radiomics[idx],
            self.labels[idx]
        )

train_dataset = MultiModalDataset(X_train, clin_train, rad_train, y_train)
test_dataset = MultiModalDataset(X_test, clin_test, rad_test, y_test)

train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=8, shuffle=False)

# =========================================================
# 5️⃣ ATTENTION BLOCK
# =========================================================

class AttentionBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.att = nn.Sequential(
            nn.Conv3d(channels, channels, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return x * self.att(x)

# =========================================================
# 6️⃣ FINAL MULTIMODAL MODEL
# =========================================================

class FinalResearchModel(nn.Module):
    def __init__(self):
        super().__init__()

        self.features = nn.Sequential(
            nn.Conv3d(1, 16, 3, padding=1),
            nn.BatchNorm3d(16),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(16, 32, 3, padding=1),
            nn.BatchNorm3d(32),
            nn.ReLU(),
            nn.MaxPool3d(2),

            nn.Conv3d(32, 64, 3, padding=1),
            nn.BatchNorm3d(64),
            nn.ReLU(),
            nn.MaxPool3d(2),
        )

        self.att = AttentionBlock(64)
        self.flatten = nn.Flatten()

        # Auto compute flatten size
        with torch.no_grad():
            dummy = torch.zeros(1, 1, X_img.shape[2], X_img.shape[3], X_img.shape[4])
            dummy = self.features(dummy)
            dummy = self.att(dummy)
            flatten_size = dummy.view(1, -1).shape[1]

        self.img_fc = nn.Linear(flatten_size, 256)

        self.clin_fc = nn.Sequential(
            nn.Linear(2, 32),
            nn.ReLU()
        )

        self.rad_fc = nn.Sequential(
            nn.Linear(7, 32),
            nn.ReLU()
        )

        self.fusion = nn.Sequential(
            nn.Linear(256 + 32 + 32, 256),
            nn.ReLU(),
            nn.Dropout(0.4)
        )

        self.output = nn.Linear(256, 1)

    def forward(self, img, clin, rad):
        x = self.features(img)
        x = self.att(x)
        x = self.flatten(x)
        x = self.img_fc(x)

        clin_feat = self.clin_fc(clin)
        rad_feat = self.rad_fc(rad)

        combined = torch.cat((x, clin_feat, rad_feat), dim=1)
        fused = self.fusion(combined)

        return self.output(fused)

# =========================================================
# 7️⃣ TRAINING LOOP
# =========================================================

model = FinalResearchModel().to(device)

pos_weight = torch.tensor([(len(y_train) - y_train.sum()) / y_train.sum()]).to(device)
criterion = nn.BCEWithLogitsLoss(pos_weight=pos_weight)

optimizer = optim.Adam(model.parameters(), lr=0.0003)

epochs = 30
best_auc = 0
patience = 5
early_stop = 0

for epoch in range(epochs):

    model.train()
    total_loss = 0

    train_bar = tqdm(train_loader, desc=f"Epoch {epoch+1}/{epochs}")

    for img, clin, rad, labels in train_bar:

        img = img.to(device)
        clin = clin.to(device)
        rad = rad.to(device)
        labels = labels.to(device)

        optimizer.zero_grad()
        outputs = model(img, clin, rad).squeeze()

        loss = criterion(outputs, labels)
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        train_bar.set_postfix(loss=loss.item())

    # Evaluation
    model.eval()
    probs = []
    labels_list = []

    with torch.no_grad():
        for img, clin, rad, labels in test_loader:

            img = img.to(device)
            clin = clin.to(device)
            rad = rad.to(device)

            outputs = model(img, clin, rad).squeeze()
            prob = torch.sigmoid(outputs)

            probs.extend(prob.cpu().numpy())
            labels_list.extend(labels.numpy())

    auc = roc_auc_score(labels_list, probs)
    acc = accuracy_score(labels_list, np.array(probs) > 0.5)

    print(f"\nEpoch {epoch+1} | Loss {total_loss:.4f} | Acc {acc:.4f} | AUC {auc:.4f}")

    if auc > best_auc:
        best_auc = auc
        torch.save(model.state_dict(), "FINAL_MULTIMODAL_MODEL.pth")
        print("🔥 Best model saved")
        early_stop = 0
    else:
        early_stop += 1

    if early_stop >= patience:
        print("🛑 Early stopping triggered")
        break

print("\n🎯 FINAL BEST AUC:", best_auc)
```

```python
# =========================================================
# 🔥 FINAL STABLE MULTIMODAL PIPELINE (PATH SAFE)
# =========================================================

import os
import torch
import torch.nn as nn
import torch.optim as optim
import numpy as np
import pandas as pd
import pydicom
import cv2
from tqdm import tqdm
from torch.utils.data import Dataset, DataLoader
from sklearn.model_selection import KFold
from sklearn.metrics import roc_auc_score, accuracy_score

device = torch.device("cpu")

NUM_SLICES = 40
IMG_SIZE = 192
BATCH_SIZE = 2
EPOCHS = 5

# =========================================================
# VERIFY DATA FOLDER
# =========================================================

print("Base Path:", base_path)
available_patients = set(os.listdir(base_path))
print("Total folders available:", len(available_patients))

# Keep only patients that actually exist
valid_patients = [p for p in label_dict.keys() if p in available_patients]

print("Patients in labels:", len(label_dict))
print("Valid patients found on disk:", len(valid_patients))

# =========================================================
# SAFE CT LOADER
# =========================================================

def load_ct(series_path):
    try:
        slices=[]
        for f in sorted(os.listdir(series_path)):
            if f.endswith(".dcm"):
                ds=pydicom.dcmread(os.path.join(series_path,f))
                slices.append(ds.pixel_array)

        if len(slices)<NUM_SLICES:
            return None

        mid=len(slices)//2
        half=NUM_SLICES//2
        slices=slices[mid-half:mid+half]

        processed=[]
        for s in slices:
            s=cv2.resize(s,(IMG_SIZE,IMG_SIZE))
            processed.append(s)

        volume=np.stack(processed)
        volume=(volume-volume.mean())/(volume.std()+1e-5)

        return volume
    except:
        return None

# =========================================================
# DATASET
# =========================================================

class MultiModalDataset(Dataset):
    def __init__(self, patient_list):
        self.patient_list = patient_list

    def __len__(self):
        return len(self.patient_list)

    def __getitem__(self, idx):

        patient = self.patient_list[idx]
        patient_path = os.path.join(base_path, patient)

        if not os.path.exists(patient_path):
            raise ValueError(f"Missing folder {patient}")

        volume=None

        for study in os.listdir(patient_path):
            study_path=os.path.join(patient_path,study)

            if not os.path.isdir(study_path):
                continue

            for series in os.listdir(study_path):
                series_path=os.path.join(study_path,series)

                if os.path.isdir(series_path):
                    volume=load_ct(series_path)
                    if volume is not None:
                        break
            if volume is not None:
                break

        if volume is None:
            raise ValueError(f"Could not load CT for {patient}")

        img=torch.tensor(volume,dtype=torch.float32).unsqueeze(0)

        clin=torch.tensor([
            clinical_df.loc[patient]['Age at Histological Diagnosis'],
            1 if clinical_df.loc[patient]['Gender']=='Male' else 0
        ],dtype=torch.float32)

        label=torch.tensor(label_dict[patient],dtype=torch.float32)

        stage_str=str(clinical_df.loc[patient]['Pathological T stage'])
        stage_map={'T1':0,'T2':1,'T3':2,'T4':3}
        stage=torch.tensor(stage_map.get(stage_str,0),dtype=torch.long)

        survival_days=clinical_df.loc[patient]['Time to Death (days)']
        if pd.isna(survival_days):
            survival_days=0
        survival=torch.tensor(float(survival_days),dtype=torch.float32)

        radiomics=np.array([
            np.mean(volume),
            np.std(volume),
            np.min(volume),
            np.max(volume),
            np.percentile(volume,25),
            np.percentile(volume,75)
        ])

        rad=torch.tensor(radiomics,dtype=torch.float32)

        return img,clin,rad,label,stage,survival

# =========================================================
# MODEL
# =========================================================

class ImageBranch(nn.Module):
    def __init__(self):
        super().__init__()
        self.net=nn.Sequential(
            nn.Conv3d(1,16,3,padding=1),nn.ReLU(),nn.MaxPool3d(2),
            nn.Conv3d(16,32,3,padding=1),nn.ReLU(),nn.MaxPool3d(2),
            nn.Conv3d(32,64,3,padding=1),nn.ReLU(),
            nn.AdaptiveAvgPool3d(1)
        )
    def forward(self,x):
        return self.net(x).view(x.size(0),-1)

class ClinicalBranch(nn.Module):
    def __init__(self):
        super().__init__()
        self.net=nn.Sequential(
            nn.Linear(2,32),nn.ReLU(),
            nn.Linear(32,32),nn.ReLU()
        )
    def forward(self,x):
        return self.net(x)

class RadiomicsBranch(nn.Module):
    def __init__(self):
        super().__init__()
        self.net=nn.Sequential(
            nn.Linear(6,32),nn.ReLU(),
            nn.Linear(32,32),nn.ReLU()
        )
    def forward(self,x):
        return self.net(x)

class FullModel(nn.Module):
    def __init__(self):
        super().__init__()
        self.img=ImageBranch()
        self.clin=ClinicalBranch()
        self.rad=RadiomicsBranch()

        self.fusion=nn.Sequential(
            nn.Linear(64+32+32,128),
            nn.ReLU(),
            nn.Dropout(0.4)
        )

        self.cancer=nn.Linear(128,1)
        self.stage=nn.Linear(128,4)
        self.survival=nn.Linear(128,1)

    def forward(self,img,clin,rad):
        f1=self.img(img)
        f2=self.clin(clin)
        f3=self.rad(rad)

        combined=torch.cat([f1,f2,f3],dim=1)
        fused=self.fusion(combined)

        return (
            self.cancer(fused),
            self.stage(fused),
            self.survival(fused)
        )

# =========================================================
# CROSS VALIDATION
# =========================================================

kf=KFold(n_splits=3,shuffle=True,random_state=42)

for fold,(train_idx,test_idx) in enumerate(kf.split(valid_patients)):

    print(f"\n===== FOLD {fold+1} =====")

    train_pat=[valid_patients[i] for i in train_idx]
    test_pat=[valid_patients[i] for i in test_idx]

    train_loader=DataLoader(MultiModalDataset(train_pat),
                            batch_size=BATCH_SIZE,shuffle=True)

    test_loader=DataLoader(MultiModalDataset(test_pat),
                           batch_size=BATCH_SIZE)

    model=FullModel().to(device)

    loss_cls=nn.BCEWithLogitsLoss()
    loss_stage=nn.CrossEntropyLoss()
    loss_surv=nn.MSELoss()

    optimizer=optim.Adam(model.parameters(),lr=0.0003)

    for epoch in range(EPOCHS):

        model.train()
        total_loss=0

        for img,clin,rad,label,stage,survival in tqdm(train_loader):

            img,clin,rad=img.to(device),clin.to(device),rad.to(device)
            label,stage,survival=label.to(device),stage.to(device),survival.to(device)

            optimizer.zero_grad()

            o1,o2,o3=model(img,clin,rad)

            l1=loss_cls(o1.squeeze(),label)
            l2=loss_stage(o2,stage)
            l3=loss_surv(o3.squeeze(),survival)

            loss=l1+0.5*l2+0.1*l3
            loss.backward()
            optimizer.step()

            total_loss+=loss.item()

        print(f"Epoch {epoch+1} Loss {total_loss:.4f}")

print("Training Complete")
```

```python
import os
import torch
import numpy as np
from tqdm import tqdm

from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.utils.class_weight import compute_class_weight

from torchvision import datasets, transforms, models
from torch.utils.data import DataLoader
from torch import nn, optim

# =========================================
# PATH
# =========================================

DATA_DIR = r"C:\lung_cancer\Data"
MODEL_PATH = r"C:\lung_cancer\lung_detection_model_v2.pth"

train_dir = os.path.join(DATA_DIR,"train")
val_dir = os.path.join(DATA_DIR,"valid")
test_dir = os.path.join(DATA_DIR,"test")

# =========================================
# DEVICE
# =========================================

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:",device)

# =========================================
# TRANSFORMS (Better Augmentation)
# =========================================

train_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.RandomHorizontalFlip(),
    transforms.RandomRotation(20),
    transforms.RandomAffine(10, translate=(0.1,0.1)),
    transforms.ColorJitter(brightness=0.2, contrast=0.2),
    transforms.ToTensor()
])

val_transform = transforms.Compose([
    transforms.Resize((224,224)),
    transforms.ToTensor()
])

# =========================================
# DATASETS
# =========================================

train_dataset = datasets.ImageFolder(train_dir,transform=train_transform)
val_dataset = datasets.ImageFolder(val_dir,transform=val_transform)
test_dataset = datasets.ImageFolder(test_dir,transform=val_transform)

print("Classes:",train_dataset.classes)

# =========================================
# DATALOADERS
# =========================================

train_loader = DataLoader(train_dataset,batch_size=4,shuffle=True)
val_loader = DataLoader(val_dataset,batch_size=4)
test_loader = DataLoader(test_dataset,batch_size=4)

# =========================================
# CLASS BALANCING
# =========================================

labels = [label for _,label in train_dataset]

class_weights = compute_class_weight(
    class_weight="balanced",
    classes=np.unique(labels),
    y=labels
)

class_weights = torch.tensor(class_weights,dtype=torch.float).to(device)

print("Class weights:",class_weights)

# =========================================
# MODEL (EfficientNet)
# =========================================

model = models.efficientnet_b0(weights="DEFAULT")

num_features = model.classifier[1].in_features

model.classifier[1] = nn.Linear(num_features,len(train_dataset.classes))

model = model.to(device)

# =========================================
# LOSS + OPTIMIZER
# =========================================

criterion = nn.CrossEntropyLoss(weight=class_weights)

optimizer = optim.Adam(model.parameters(),lr=0.0001)

# =========================================
# TRAINING
# =========================================

EPOCHS = 15
best_acc = 0

for epoch in range(EPOCHS):

    print(f"\nEpoch {epoch+1}/{EPOCHS}")

    model.train()

    preds=[]
    labels=[]

    for images,target in tqdm(train_loader):

        images=images.to(device)
        target=target.to(device)

        optimizer.zero_grad()

        outputs=model(images)

        loss=criterion(outputs,target)

        loss.backward()

        optimizer.step()

        pred=torch.argmax(outputs,1)

        preds.extend(pred.cpu().numpy())
        labels.extend(target.cpu().numpy())

    train_acc = accuracy_score(labels,preds)

    print("Train Accuracy:",train_acc)

    # =====================================
    # VALIDATION
    # =====================================

    model.eval()

    preds=[]
    labels=[]

    with torch.no_grad():

        for images,target in val_loader:

            images=images.to(device)
            target=target.to(device)

            outputs=model(images)

            pred=torch.argmax(outputs,1)

            preds.extend(pred.cpu().numpy())
            labels.extend(target.cpu().numpy())

    val_acc = accuracy_score(labels,preds)

    print("Validation Accuracy:",val_acc)

    if val_acc>best_acc:

        best_acc=val_acc

        torch.save(model.state_dict(),MODEL_PATH)

        print("🔥 Model Saved")

# =========================================
# TESTING
# =========================================

print("\nTesting Best Model")

model.load_state_dict(torch.load(MODEL_PATH))

model.eval()

preds=[]
labels=[]

with torch.no_grad():

    for images,target in tqdm(test_loader):

        images=images.to(device)
        target=target.to(device)

        outputs=model(images)

        pred=torch.argmax(outputs,1)

        preds.extend(pred.cpu().numpy())
        labels.extend(target.cpu().numpy())

test_acc = accuracy_score(labels,preds)

print("\nFinal Test Accuracy:",test_acc)

cm = confusion_matrix(labels,preds)

print("\nConfusion Matrix:")
print(cm)

print("\nModel saved at:")
print(MODEL_PATH)
```
